```python
# Rating missing hai toh median se fill kar rhe hai

df['rating'] = df['rating'].fillna(df['rating'].median())
```

```python
# jaha discount nhi diya hai wha 0 kr diye hai

df['discount'] = df['discount'].fillna(0)
```

```python
# Brand missing hai toh 'Unknown' daal de rhe hai

df['brand'] = df['brand'].fillna('Unknown')
```

```python
print(df.isnull().sum())
```

```python
# ab koi null value nhi bachi
```

```python
df['original_price'] = np.where(
    df['discount'] == 0,  
    df['price'],
    (df['price'] / (1 - df['discount'] / 100))
)
```

```python
df['original_price'] = df['original_price'].round(2)
```

```python
print(df[['price', 'discount', 'original_price']].head(10))
```

```python
print(df.columns) 
```

```python
print(df.isnull().sum())
```

```python
# koi null values nhi bachi hai
```

```python
# brand mein jaha jaha Brand: alag se likha hai usko hatana hai

df['brand'] = df['brand'].str.replace('Brand: ', '', regex=False).str.strip()
```

```python
print(df['brand'].value_counts())
```

```python
print(df.dtypes)
```

```python
# min 5000 se neeche koi smartphone nhi hoga ye consider krke chal rhe hai

df = df[df['price'] > 5000]
```

```python
df['site'] = 'Amazon'
```
