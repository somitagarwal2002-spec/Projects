```python
import pandas as pd
import numpy as np
```

```python
def clean_amazon_data(file_path):
    df = pd.read_csv(file_path, skiprows=3)
    print(f"Loaded! Shape: {df.shape}")

    required_columns = ['title', 'price', 'rating', 'reviews', 'discount', 'brand', 'product_url']
    missing_cols = [col for col in required_columns if col not in df.columns]

    if missing_cols:
        print(f"Error: This column is missing: {missing_cols}")
        return None
    print("All required columns available")

    before = df.shape[0]
    df = df.drop_duplicates()
    after = df.shape[0]
    print(f"Duplicates removed: {before - after} rows")

    # Cleaning Data

    df['rating']   = df['rating'].fillna(df['rating'].median())
    df['discount'] = df['discount'].fillna(0)
    df['brand']    = df['brand'].fillna('Unknown')

    # Agar kisi column ka data type nhi shi hua to 

    df['price']    = pd.to_numeric(df['price'],    errors='coerce')
    df['discount'] = pd.to_numeric(df['discount'], errors='coerce').fillna(0)
    df['reviews']  = pd.to_numeric(df['reviews'],  errors='coerce').fillna(0).astype(int)

    # Brand column cleaning

    df['brand'] = df['brand'].str.replace('Brand: ', '', regex=False).str.strip()

    # Adding Original Price Column
    
    df['original_price'] = np.where(
        df['discount'] == 0,
        df['price'],
        (df['price'] / (1 - df['discount'] / 100))).round(2)
    print("original_price column Added")

    df['site'] = "Amazon"

    output_path = file_path.replace('.csv', '_cleaned.csv')
    df.to_csv(output_path, index=False)

    print(f"\nDone! Cleaned File is Saved: {output_path}")
    print(f"Final Shape: {df.shape}")
    
    return df
```
